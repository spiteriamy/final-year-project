## **Load OOS Data**

Out-of-Scope sentences taken from CLINC150 dataset.  
https://github.com/clinc/oos-eval.git

```python
!git clone https://github.com/clinc/oos-eval.git
```

```python
data_path = os.path.join('oos-eval', 'data')
```

```python
with open(os.path.join(data_path, 'data_full.json'), 'r') as f:
  data = json.load(f)
```

```python
all_sentences = []
for key, val in data.items():
  for entry in val:
    all_sentences.append(entry[0])
```

## **Load Intent Classification Data**

```python
os.chdir('/content/drive/MyDrive/FYP')
```

```python
DATA_PATH = os.path.join("intent_data.csv")
```

```python
# load necessary columns into df
df = pd.read_csv(DATA_PATH, usecols=['question', 'intent'])
df.head()
```

```python
# dataset summary
print(f"Dataset size: {len(df)} rows")
print(f"Intents: {df['intent'].value_counts().to_dict()}")
```

## **Extend Dataset with OOS Sentences**

```python
# select subset of sentences to be added
n = 70
sentences = random.sample(all_sentences, n)
```

```python
# Create a new dataframe from the list
new_df = pd.DataFrame({
    'question': sentences,
    'intent': ['oos'] * len(sentences)
})

new_df.head()
```

```python
# Append oos sentences to original dataframe
df = pd.concat([df, new_df], ignore_index=True)
```

```python
print(df.tail())
print(df['intent'].value_counts())
```

```python
# save to file
df.to_csv("data_with_oos.csv", index=False)
```
